## Chinese Character Lookup

This notebook looks up Chinese characters or words in a dictionary and displays a component decomposition for each character. It uses `hanzipy` for dictionary and decomposition data, and `ipywidgets` to create an interactive input field and button.

```python
from hanzipy.dictionary import HanziDictionary
from hanzipy.decomposer import HanziDecomposer
```

### Import the Hanzi tools

Import the dictionary and decomposition classes from `hanzipy`. The dictionary provides definitions and pronunciation information; the decomposer breaks a character into its components.


```python
dictionary = HanziDictionary()
decomposer = HanziDecomposer()
```

### Create the Hanzi tools

Create one dictionary object and one decomposer object. These are reused whenever a lookup is performed, rather than recreated for every button click.


### Import the notebook widget tools

Import `ipywidgets` to build the text box and button. `display` shows the widgets in the notebook, while `clear_output` clears the previous lookup results before showing new ones.


```python
import ipywidgets as widgets
from IPython.display import display, clear_output
```

### Set up the interface

Create a text box with `学` as its initial value, a “Look up” button, and an output area for results. The placeholder and description guide the user on what to enter.

```python
query = widgets.Text(
    value="学",
    placeholder="Enter a Chinese character or word",
    description="Hanzi:",
)
button = widgets.Button(description="Look up")
output = widgets.Output()
```

### Define the lookup function

When the button is clicked, this function reads and trims the text-box input, then clears the previous results. If the input is blank, it displays a prompt and stops.

It then uses `hanzipy` in two ways:

- **Dictionary lookup:** `dictionary.definition_lookup(text)` returns matching dictionary entries, with traditional and simplified forms, pinyin, and definitions. A character can have multiple entries, such as different readings or senses, so each result is printed separately.
- **Character decomposition:** `decomposer.decompose(character)` reports components for each character in the input. The function processes the text one character at a time. This is a structural breakdown; it is not a translation or an explanation of what each component means.

The function also handles lookup or decomposition errors so one failure is reported in the output instead of stopping the whole lookup.


```python
def look_up(_):
    text = query.value.strip()

    with output:
        clear_output()

        if not text:
            print("Enter a character or word first.")
            return

        # hanzipy's dictionary lookup returns matching entries, including
        # traditional/simplified forms, pinyin, and definitions.
        print("Dictionary entries")
        print("Each result is a possible dictionary entry; a character may have multiple readings.")
        try:
            # https://github.com/Synkied/hanzipy#dictionarydefinition_lookupcharacterword-script_typenone
            entries = dictionary.definition_lookup(text)
        except Exception as error:
            entries = None
            print(f"Dictionary lookup failed: {error}")

        if entries:
            for number, entry in enumerate(entries, start=1):
                # for entry content, see hanzipy docs
                traditional = entry.get("traditional", "")
                simplified = entry.get("simplified", "")
                pinyin = entry.get("pinyin", "")
                definition = entry.get("definition", "")

                print(f"{number}. {traditional} / {simplified}")
                print(f"   Pinyin: {pinyin or 'not provided'}")
                print(f"   Definition: {definition or 'not provided'}")
        elif entries is not None:
            print("No dictionary entries found.")

        # decomposer reports components for each individual character.
        # This is a structural breakdown, not a word definition or translation.
        print("\nCharacter components")

        for character in text:
            try:
                components = decomposer.decompose(character)
                print(f"{character}: {components}")
            except Exception as error:
                print(f"{character}: decomposition failed ({error})")
```

### Connect and display the interface

Connect the button to `look_up`, so clicking it runs the lookup function. Then display the text box and button side by side, followed by the output area where the results appear.

```python
button.on_click(look_up)
display(widgets.HBox([query, button]), output)
```
